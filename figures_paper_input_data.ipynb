```python
import xarray as xr
import pygmt
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import import_ipynb

# net cdf files
# power_percentage_w1 = xr.open_dataset('world_5km_output_sumpower_w1_proportion_merged_v0.nc', 
#                                       chunks={'period': 10})
# power_percentage_w2 = xr.open_dataset('world_5km_output_sumpower_w2_proportions_merged_v0.nc', 
#                                       chunks={'period': 10})
power_w1 = xr.open_dataset('world_5km_output_sumpower_w1_merged_v0.nc', 
                           chunks={'period': 10})
power_w2 = xr.open_dataset('world_5km_output_sumpower_w2_merged_v0.nc', 
                           chunks={'period': 10})

# # see the periods in the dataset
# print(power_percentage_w1.period.values)
# # number of periods
# print(f'Number of periods: {len(power_percentage_w1.period.values)}')
```

```python
power_percentage_w1 = power_w1.copy()
power_percentage_w2 = power_w2.copy()

# calculate the percentage of power at each period
total_power_w1 = power_w1['rect_power'].sum(dim='period')
power_percentage_w1['proportion'] = (power_w1['rect_power'] / total_power_w1) * 100

total_power_w2 = power_w2['rect_power'].sum(dim='period')
power_percentage_w2['proportion'] = (power_w2['rect_power'] / total_power_w2) * 100
```

```python
# now just isolating the different periods I'm interested in

### WINDOW 1 - Actual power
triyearly_w1 = power_w1.rect_power.sel(period=slice(115.1,123.4)).sum(dim="period", skipna=False)
biyearly_w1 = power_w1.rect_power.sel(period=slice(174.5, 187.)).sum(dim="period", skipna=False)
yearly_w1 = power_w1.rect_power.sel(period=slice(349, 374.)).sum(dim="period", skipna=False)
twoyear_w1 = power_w1.rect_power.sel(period=slice(697.9 , 748)).sum(dim="period", skipna=False)

### WINDOW 2 - Actual power
triyearly_w2 = power_w2.rect_power.sel(period=slice(115.1,123.4)).sum(dim="period", skipna=False)
biyearly_w2 = power_w2.rect_power.sel(period=slice(174.5, 187.)).sum(dim="period", skipna=False)
yearly_w2 = power_w2.rect_power.sel(period=slice(349, 374.)).sum(dim="period", skipna=False)
twoyear_w2 = power_w2.rect_power.sel(period=slice(697.9, 748)).sum(dim="period", skipna=False)

# Save these datasets in two dictionaries so easier to plot some stuff later
power_datasets_w1  = {
    '4 months': triyearly_w1,
    '6 months': biyearly_w1,
    '1 year': yearly_w1,
    '2 years': twoyear_w1,

}

power_datasets_w2  = {
    '4 months': triyearly_w2,
    '6 months': biyearly_w2,
    '1 year': yearly_w2,
    '2 years': twoyear_w2,
}

# differences between windows
# define DIFF dictionary
power_diff_dict = {}

for key in power_datasets_w1:
    # define first and second dataset
    w1 = power_datasets_w1[key]
    w2 = power_datasets_w2[key]
    dataset_diff = w2 - w1
    power_diff_dict[key] = dataset_diff
```
